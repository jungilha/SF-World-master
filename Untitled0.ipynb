```python
import re, numpy as np
import pandas as pd

# 존재하는 열만 드롭(둘 중 어딘가에 없으면 자동 건너뜀)
drop_cols = ['ID', 'Name', 'New_Price', 'Price']

X = train.drop(columns=[c for c in drop_cols if c in train.columns]).copy()
X_test = test.drop(columns=[c for c in drop_cols if c in test.columns]).copy()

y = train['Price'].values  # 타깃 유지
print('X cols:', X.columns.tolist())
print('X_test cols:', X_test.columns.tolist())
```

```python
def to_float(s):
    if pd.isna(s):
        return np.nan
    s = str(s).lower().replace(',', ' ')
    # 숫자만 추출 (소수 포함)
    m = re.findall(r'[\d.]+', s)
    if not m:
        return np.nan
    return float(m[0])

# 문자열이지만 숫자 정보가 들어있는 열들
num_like_cols = ['Mileage','Engine','Power','Seats','No. of Doors',
                 'Kilometers_Driven','Year']

for col in num_like_cols:
    if col in X.columns:
        X[col] = X[col].apply(to_float)
    if col in X_test.columns:
        X_test[col] = X_test[col].apply(to_float)
```

```python
num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = [c for c in X.columns if c not in num_cols]

print('Numeric:', num_cols)
print('Categorical:', cat_cols)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

numeric_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
])

categorical_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore')),
])

preprocess = ColumnTransformer([
    ('num', numeric_tf, num_cols),
    ('cat', categorical_tf, cat_cols),
])

model = RandomForestRegressor(
    n_estimators=600, max_depth=None,
    random_state=42, n_jobs=-1
)

pipe = Pipeline([
    ('preprocess', preprocess),
    ('model', model)
])
```

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import make_scorer
import numpy as np

def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / np.clip(y_true, 1e-3, None))) * 100

mape_scorer = make_scorer(mape, greater_is_better=False)
cv = KFold(n_splits=5, shuffle=True, random_state=42)

scores = -cross_val_score(pipe, X, y, cv=cv, scoring=mape_scorer, n_jobs=-1)
print(f"CV MAPE: {scores.mean():.2f}% ± {scores.std():.2f}%")
```

```python
pipe.fit(X, y)
pred = pipe.predict(X_test)

sub = pd.DataFrame({
    'ID': test['ID'] if 'ID' in test.columns else np.arange(len(test)),
    'Price': pred
})
sub.to_csv('submission.csv', index=False)
sub.head()
```
